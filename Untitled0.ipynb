```python
import pandas as pd

data = {
    'Employee_ID': [101, 102, 103, 104, 105, 105, 106, 107],
    'Name': ['Rahul', 'Priya', 'Amit', 'Sneha', None, None, 'Karan', 'Neha'],
    'Age': [25, 28, None, 30, 26, 26, 29, 32],
    'Department': ['IT', 'HR', 'it', 'Finance', 'HR', 'HR', 'IT ', 'finance'],
    'Salary': ['30000', '45000', '35000', None, '40000', '40000', '50000', '55000'],
    'Joining_Date': [
        '2024-01-10', '2023-05-15', '2024-03-20',
        '2022-08-12', '2023-11-01', '2023-11-01',
        '2024-02-10', '2022-06-15'
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())

df['Salary'] = df['Salary'].fillna(
    pd.to_numeric(df['Salary'], errors='coerce').median()
)

df['Name'] = df['Name'].fillna('Unknown')
```

```python
df = df.drop_duplicates()
```

```python
df['Department'] = (
    df['Department']
    .str.strip()
    .str.title()
)
```

```python
df['Salary'] = (
    df['Salary']
    .astype(str)
    .str.replace(',', '', regex=False)
)

df['Salary'] = pd.to_numeric(df['Salary'], errors='coerce')
```

```python
df['Joining_Date'] = pd.to_datetime(
    df['Joining_Date'],
    errors='coerce'
)
```

```python
df = df.reset_index(drop=True)
```
